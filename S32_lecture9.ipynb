### The geometry of the strange Lorenz attractor

You can't really get a feel for the geometry of the Lorenz attractor from a 2-D static plot, so here is an animation. You may think of the wings of the butterfly as 2D: Given the chaotic nature we will eventually fill the entire disk of these wings and so we need two dimensions to describe every point on these disks. But if the attractor was truely 2D then it wouldn't actually be chaotic (recall the Poincaré-Bendixson theorem!) and in fact where the two wings meet the attractor does have properties that remind more of 3D. Lines of the trajectories don't just lie next to another in some 2D plane but also lie over another in a third dimension, so visually appearing to "cross" each other. In this part of the attractor we do need a third dimension to describe the behaviour of the Lorenz system fully. So clearly the dimension is more than 2D, but is it actually 3D or something in between?

We will now explore the concept of fractal dimension (and different ways to estimate it), starting in simpler systems. To estimate the dimensionality of the Lorenz attractor, we need a method like the box counting algorithm introduced below, which, it turns out, estimates its dimension to be 2.05... (typically with some small error), so indeed _fractal_!

#### 8.2: The concept of attractor dimension

When we introduced the Lorenz system, we argued that it’s chaotic dynamics meant that (for some parameter values) it had to evolve on an attractor with a finite but non-integer, or “fractal”, dimension. In contrast, stochastic and Hamiltonian systems generate space-filling trajectories with an integer (or infinite) dimension.

If an unknown system appears to evolve on an attractor with a finite but non-integer dimension, this would seem to be a strong indication that it is chaotic, that is: 

- *deterministic:* (how else would it remain in such a confined region of phase-space?), 
- *aperiodic* (trajectories never touch on a strange attractor) and 
- displays *sensitive dependence* on initial conditions.

This last point is less obvious, and a general proof that a fractal attractor dimension implies sensitive dependence of trajectories on initial conditions is hard. So hard, in fact, that people these days tend to focus on predictability itself as the hall-mark of chaos, not attractor dimension. But a method of identifying chaotic behaviour purely from process output clearly has its attractions. So how can we estimate the dimension of an attractor without knowing the underlying process dynamics?

### Similarity dimension

We can define the dimension of an object in terms of how the number of scaled-down copies of that object that are 
required to cover the original varies with the scaling factor. Think about

- 1D: A line segment. We can take a 2x smaller line segment and use 2 of those to cover the original line segment again. Dimensionality would be 1.
- 2D: An area, e.g. a square. We can take a 2x smaller square but now we need 4 of those to cover the original square again. Dimensionality is therefore 2.
- 3D: A volume, e.g. a cube. We can take a 2x smaller cube but now we need 9 of those to cover the original cube. Dimensionality is therefore 3.

#### Generalisation of this similarity dimension

Lets work out the dimension of the whiteboard. If $m$ is the number of copies and $r$ is the scaling factor, then
the so-called ''similarity dimension'' $d$ is defined by

$$ m = r^d $$ 
or 
$$ d = \frac{\log m}{\log r} $$

### The Cantor Set
This is defined by taking the set of points between 0 and 1, removing the middle third, and then repeating the exercise indefinitely.  If we scale down the Cantor set by three, we need two copies to cover the original, so $m=2$ when $r=3$ and
$$ d = \frac{\log 2}{\log 3} \simeq 0.63 $$

### The Koch snowflake

The Koch snowflake is constructed by starting with an equilateral triangle and dividing every of the three faces into three section of which the middle one is replaced with another equilateral triangle. For each of the new 12 faces we repeat: Divide into three sections and replace the middle one with an equilateral triangle

![](https://upload.wikimedia.org/wikipedia/commons/f/fd/Von_Koch_curve.gif)

What is the (fractal) similarity dimension of the snowflake?

We observe

- Every new triangle is shrunk by 3x compared to the previous one
- But every face turns into 4 faces during this process

So $m = 4$ and $r=3$ which leaves us with

$$
d = \frac{\log{4}}{\log{3}} = 1.26...
$$

And the similarity dimension of the snowflake is 1.26...! In other words as the length of the outline of the snowflake goes to infinity (you can verify that!) but the area is clearly bounded (given it will never exceeds its embedding square) we are dealing with a geometric object here that does not simply behave like a line, especially because the infinite length of its outline is _uncountable_.

### Box counting dimensions

For non-self-similar objects (like the Lorenz attractor), we need a more general method. Imagine you have some data and you can't easily figure out the number of copies $m$ or the shrinking $r$ because shapes do not exactly repeat themselves when you zoom in. This is the case for many real world fractals. The simplest approach to quantify a fractal dimension in this case is the *box-counting dimension*. This is defined by how the number of boxes $N(\epsilon)$ required to cover that object scales with the size (edge-length, $\epsilon$) of the boxes.  For a line of length $L$, $N(\epsilon) = L/\epsilon$, for a surface of area $A$, $N(\epsilon) = A/\epsilon^2$ and so on. In general

$$ N(\epsilon) \propto 1/\epsilon^d $$ so
$$ d = \lim_{\epsilon \rightarrow 0} 
\frac{\log N(\epsilon)}{\log(1/\epsilon)} $$

if that limit exists. 

Here is an example of applying this to the coastline of Great Britain 

![](https://upload.wikimedia.org/wikipedia/commons/2/28/Great_Britain_Box.svg)

(By Prokofiev - Own work, CC BY-SA 3.0, https://commons.wikimedia.org/w/index.php?curid=12042116)

Now the question is how many boxes do we need to cover the coastline as the box size decreases? If the dimension of the coastline would be 1D then a smaller and smaller share of all boxes covering the whole embeddeding 2D space would be needed. If the coastline would indeed be 2D (which would be, well, an area) then the share of boxes would approach some limit. With the real coastline it turns out the answer is somewhere in between and the number of boxes needed to cover the coastline yields a dimension of 1.2 or so. So the coast"line" of Great Britain is actually fractal and, so is topologically so complex that it is between a line and an area!

Note that there's a difference when asking for the dimensionality of the coastline versus the dimensionality of the area of Great Britain. The area continues to be 2D, because you need 2 dimensions (like 2 coordinates, longitude, latitude) to reach every point within Great Britain. And the area is finite too, using smaller and smaller boxes and counting up the covered area (not coastline!) would converge towards the true area. But the topology of the coastline itself is different, it's infinite in length and indeed _uncountably_ infinite because for smaller and smaller boxes (or rulers) you always add new segments in between coarser segments -- like we did in the Koch snowflake!
